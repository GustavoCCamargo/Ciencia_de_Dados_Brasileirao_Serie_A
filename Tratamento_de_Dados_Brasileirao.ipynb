## Tratamento de valores faltantes

```python
# Contar quantos valores faltantes existem em cada coluna
print("Quantidade de valores faltantes por coluna:")
print(df_fut.isnull().sum())

# Mostrar a porcentagem de valores faltantes por coluna
print("\nPorcentagem de valores faltantes por coluna:")
print((df_fut.isnull().mean() * 100).round(2))

# Visualizar quais linhas têm valores faltantes
print("\nLinhas com pelo menos um valor faltante:")
print(df_fut[df_fut.isnull().any(axis=1)])
```

```python
colunas_estatisticas = [
    'escanteios_mandante', 'escanteios_visitante', 'faltas_mandante', 'faltas_visitante',
    'chutes_bola_parada_mandante', 'chutes_bola_parada_visitante',
    'defesas_mandante', 'defesas_visitante',
    'impedimentos_mandante', 'impedimentos_visitante',
    'chutes_mandante', 'chutes_visitante',
    'chutes_fora_mandante', 'chutes_fora_visitante',
    'gols_1_tempo_visitante', 'gols_1_tempo_mandante', "gols_mandante", "gols_visitante"
]

import pandas as pd
import numpy as np

colunas_estatisticas = [
    'escanteios_mandante', 'escanteios_visitante', 'faltas_mandante', 'faltas_visitante',
    'chutes_bola_parada_mandante', 'chutes_bola_parada_visitante',
    'defesas_mandante', 'defesas_visitante',
    'impedimentos_mandante', 'impedimentos_visitante',
    'chutes_mandante', 'chutes_visitante',
    'chutes_fora_mandante', 'chutes_fora_visitante',
    'gols_1_tempo_visitante', 'gols_1_tempo_mandante', "gols_mandante", "gols_visitante"
]

for i in range(len(df_fut)):
    linha = df_fut.iloc[i]
    time_mandante = linha['time_mandante']
    time_visitante = linha['time_visitante']
    ano = linha['ano_campeonato'] # Pega o ano da linha atual

    colunas_mandante = [
        'escanteios_mandante', 'faltas_mandante', 'chutes_bola_parada_mandante',
        'defesas_mandante', 'impedimentos_mandante', 'chutes_mandante', 'chutes_fora_mandante', "gols_mandante", "gols_1_tempo_mandante"
    ]

    colunas_visitante = [
        'escanteios_visitante', 'faltas_visitante', 'chutes_bola_parada_visitante',
        'defesas_visitante', 'impedimentos_visitante', 'chutes_visitante', 'chutes_fora_visitante', "gols_visitante", "gols_1_tempo_visitante"
    ]

    # Corrige mandante
    for col in colunas_mandante:
        if pd.isna(linha[col]):
            # Filtra por time E por ano
            mandantes = df_fut[(df_fut['time_mandante'] == time_mandante) & (df_fut['ano_campeonato'] == ano)][col]
            mandantes_sem_zeros = mandantes.replace(0, np.nan)
            media = mandantes_sem_zeros.mean()

            if not np.isnan(media):
                df_fut.iloc[i, df_fut.columns.get_loc(col)] = int(round(media))

    # Corrige visitante
    for col in colunas_visitante:
        if pd.isna(linha[col]):
            # Filtra por time E por ano
            visitantes = df_fut[(df_fut['time_visitante'] == time_visitante) & (df_fut['ano_campeonato'] == ano)][col]
            visitantes_sem_zeros = visitantes.replace(0, np.nan)
            media = visitantes_sem_zeros.mean()

            if not np.isnan(media):
                df_fut.iloc[i, df_fut.columns.get_loc(col)] = int(round(media))


df_fut['publico_max'] = df_fut['publico_max'].fillna(
    df_fut.groupby('estadio')['publico_max'].transform('mean'))

df_fut['publico_max'] = df_fut['publico_max'].fillna(df_fut['publico_max'].mean())


df_fut['tecnico_mandante'] = df_fut['tecnico_mandante'].fillna('Desconhecido')
df_fut['tecnico_visitante'] = df_fut['tecnico_visitante'].fillna('Desconhecido')

df_fut['idade_media_titular_mandante'] = df_fut['idade_media_titular_mandante'].fillna(
    df_fut.groupby('time_mandante')['idade_media_titular_mandante'].transform('mean')
)

df_fut['idade_media_titular_visitante'] = df_fut['idade_media_titular_visitante'].fillna(
    df_fut.groupby('time_visitante')['idade_media_titular_visitante'].transform('mean')
)

df_fut['valor_equipe_titular_mandante'] = df_fut['valor_equipe_titular_mandante'].fillna(
    df_fut.groupby('time_mandante')['valor_equipe_titular_mandante'].transform('mean')
)

df_fut['valor_equipe_titular_visitante'] = df_fut['valor_equipe_titular_visitante'].fillna(
    df_fut.groupby('time_visitante')['valor_equipe_titular_visitante'].transform('mean')
)
```

## Tratamento de Outliers

```python
colunas_estatisticas = [
    'escanteios_mandante', 'escanteios_visitante', 'faltas_mandante', 'faltas_visitante',
    'chutes_bola_parada_mandante', 'chutes_bola_parada_visitante',
    'defesas_mandante', 'defesas_visitante',
    'impedimentos_mandante', 'impedimentos_visitante',
    'chutes_mandante', 'chutes_visitante',
    'chutes_fora_mandante', 'chutes_fora_visitante'
]

for col in colunas_estatisticas:
    q1 = df_fut[col].quantile(0.25)
    q3 = df_fut[col].quantile(0.75)
    iqr = q3 - q1
    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    outliers = df_fut[(df_fut[col] < limite_inferior) | (df_fut[col] > limite_superior)]


    if not outliers.empty: # verifica se o outlier não está vazio, empty diz que está e o not reverte a condição.

           print(f"Outliers na coluna {col}: {len(outliers)} linhas")
           plt.figure() # DEixa gráfico isolado
           plt.boxplot(df_fut[col])
           plt.title(f'Boxplot – {col}')
           plt.ylabel(col)
           plt.grid(True, linestyle='--', alpha=0.5)
           plt.show()




```

```python
# Define o limite máximo aceitável para faltas (exemplo 50, que já é um número alto)
limite_max = 50
limite_min = 0

# Substitui valores maiores que limite_max por NaN nas colunas de faltas
df_fut.loc[df_fut['faltas_mandante'] > limite_max, 'faltas_mandante'] = pd.NA
df_fut.loc[df_fut['faltas_visitante'] > limite_max, 'faltas_visitante'] = pd.NA

# Substitui valores maiores que limite_max por NaN nas colunas de faltas
df_fut.loc[df_fut['faltas_mandante'] < limite_min, 'faltas_mandante'] = pd.NA
df_fut.loc[df_fut['faltas_visitante'] < limite_min, 'faltas_visitante'] = pd.NA

# Para faltas_mandante, preenche pela média de faltas do time mandante
df_fut['faltas_mandante'] = df_fut['faltas_mandante'].fillna(
    df_fut.groupby('time_mandante')['faltas_mandante'].transform('mean')
)

df_fut['faltas_visitante'] = df_fut['faltas_visitante'].fillna(
    df_fut.groupby('time_visitante')['faltas_visitante'].transform('mean')
)
```

## Transformação de dados

```python
colunas_estatisticas = [
    'escanteios_mandante', 'escanteios_visitante', 'faltas_mandante', 'faltas_visitante',
    'chutes_bola_parada_mandante', 'chutes_bola_parada_visitante',
    'defesas_mandante', 'defesas_visitante',
    'impedimentos_mandante', 'impedimentos_visitante',
    'chutes_mandante', 'chutes_visitante',
    'chutes_fora_mandante', 'chutes_fora_visitante', 'publico_max', 'publico', 'gols_1_tempo_mandante', 'gols_1_tempo_visitante', 'gols_mandante', 'gols_visitante',
    'colocacao_mandante', 'colocacao_visitante'
]

for col in colunas_estatisticas:
    df_fut[col] = df_fut[col].astype(int)
```

## Log-Transformação

```python
# Função log-transformação (com +1 para evitar log(0))
def log_transform(x):
    return math.log(x + 1)

df_fut['log_publico'] = df_fut['publico'].apply(log_transform)
df_fut['log_publico_max'] = df_fut['publico_max'].apply(log_transform)
```

```python
df_fut = df_fut.sort_values('ano_campeonato').reset_index(drop=True)
```

```python


for i in range(len(df_fut)):
    linha = df_fut.iloc[i]
    qtd_zeros = 0

    # Adicionada a variável 'ano'
    ano = linha['ano_campeonato']

    colunas_suspeitas = [
        'escanteios_mandante', 'escanteios_visitante', 'faltas_mandante', 'faltas_visitante',
        'chutes_bola_parada_mandante', 'chutes_bola_parada_visitante', 'defesas_mandante',
        'defesas_visitante', 'impedimentos_mandante', 'impedimentos_visitante',
        'chutes_mandante', 'chutes_visitante', 'chutes_fora_mandante', 'chutes_fora_visitante'
    ]

    for col in colunas_suspeitas:
        if linha[col] == 0:
            qtd_zeros += 1

    if qtd_zeros >= 8:
        time_mandante = linha['time_mandante']
        time_visitante = linha['time_visitante']

        colunas_mandante = [
            'escanteios_mandante', 'faltas_mandante', 'chutes_bola_parada_mandante',
            'defesas_mandante', 'impedimentos_mandante', 'chutes_mandante', 'chutes_fora_mandante'
        ]

        colunas_visitante = [
            'escanteios_visitante', 'faltas_visitante', 'chutes_bola_parada_visitante',
            'defesas_visitante', 'impedimentos_visitante', 'chutes_visitante', 'chutes_fora_visitante'
        ]

        # Corrige mandante
        for col in colunas_mandante:
            if linha[col] == 0:
                # Seleciona as linhas com o mesmo time mandante e o mesmo ano
                mandantes = df_fut[(df_fut['time_mandante'] == time_mandante) & (df_fut['ano_campeonato'] == ano)][col]
                mandantes_sem_zeros = mandantes.replace(0, np.nan)
                media = mandantes_sem_zeros.mean()

                if not np.isnan(media):
                    df_fut.iloc[i, df_fut.columns.get_loc(col)] = int(round(media))

        # Corrige visitante
        for col in colunas_visitante:
            if linha[col] == 0:
                # Seleciona as linhas com o mesmo time visitante e o mesmo ano
                visitantes = df_fut[(df_fut['time_visitante'] == time_visitante) & (df_fut['ano_campeonato'] == ano)][col]
                visitantes_sem_zeros = visitantes.replace(0, np.nan)
                media = visitantes_sem_zeros.mean()

                if not np.isnan(media):
                    df_fut.iloc[i, df_fut.columns.get_loc(col)] = int(round(media))
```
